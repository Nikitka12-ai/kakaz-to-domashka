Пункт 3

```python
df = pd.read_csv('/content/drive/MyDrive/paddydataset.csv')
print("Инфа о датасете:")
df.info()
```

Пункт 4 строки и столбцы

```python
print(f"\nКоличество столбцов: {len(df.columns)}")
print("Названия столбцов:")
print(df.columns)
print(f"\nКоличество строк: {len(df)}")
print("Индексы строк:")
print(df.index)
```

Пункт 5 Замена категориальных данных количественными

```python
categorical_columns = df.select_dtypes(include=['object']).columns.tolist()
print(f"\n ДО ЗАМЕНЫ:")
print(f"Найдено категориальных столбцов: {len(categorical_columns)}")
print(f"Имена столбцов: {categorical_columns}")
if len(categorical_columns) > 0:
    print("\nПервые 5 строк ДО замены:")
    print(df[categorical_columns].head())
if len(categorical_columns) > 0:
    print(" Замена категориальных данных")
    for col in categorical_columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])
        print(f"\n  ✓ Столбец '{col}' закодирован:")
print("ПОСЛЕ ЗАМЕНЫ")
if len(categorical_columns) > 0:
    print(f"\nПервые 5 строк ПОСЛЕ замены:")
    print(df[categorical_columns].head())
```

Пункт 6

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Paddy yield(in Kg)', bins=30, kde=False, color='coral', alpha=0.7)
plt.title('Распределение урожайности риса', fontsize=16, fontweight='bold')
plt.xlabel('Урожайность (кг)', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Seedrate(in Kg)', bins=30, kde=False, color='skyblue', alpha=0.7)
plt.title('Распределение нормы высева семян', fontsize=16, fontweight='bold')
plt.xlabel('Норма высева (кг)', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Urea_40Days', bins=30, kde=False, color='lightgreen', alpha=0.7)
plt.title('Распределение внесения мочевины на 40-й день', fontsize=16, fontweight='bold')
plt.xlabel('Мочевина (кг)', fontsize=12)
plt.ylabel('Частота', fontsize=12)
plt.tight_layout()
plt.show()
```
